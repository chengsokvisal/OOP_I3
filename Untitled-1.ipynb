TP01.3.Asterisks

A.

```python
for i in range(6):
  if i==0 or i==5:
    print("*"*25)
  else:
    print("*                       *")
```

B.

```python
row = 5
for i in range(1, row+1):
  space = " " * (row-i)
  star = "*" * (2*i-1)
  print(space+star)
```

C.

```python
row = 5
for i in range(6):
  print("*" * (i*2-1))
```

D.

```python
row = 5
for i in range(1, row+1):
  space = " " * (2*(row-i))
  star = "*" * (2*i-1)
  print(space+star)
```

E.

```python
row = 3
for i in range(1, row+1):
  space = " " * (2*(row-i))
  star = "*" * (2*i-1)
  print(space+star)
for i in range(row-1, 0, -1):
  spaces = " " * (2*(row-i))
  stars = "*" * (2*i-1)
  print(spaces+stars)
```

F.

```python
row = 3
for i in range(1, row+1):
  space = " " * (row-i) 
  star = "*" * (2*i-1)
  print(space+star)
for i in range(row-1, 0, -1):
    spaces = " " * (row-i)
    stars = "*" * (2*i-1)
    print(spaces+stars)
```

G.

```python
row = 3
for i in range(row, 0, -1):
  space = " " * (row -i)
  star = "*" * (2*i-1)
  print(space+star)
for i in range(2, row+1):
  spaces = " " * (row -i)
  stars = "*" * (2*i-1)
  print(spaces+stars)
```

H.

```python
row = 3
for i in range(1, row+1):
  spaces = " " * (row-i)
  if i==1:
    print(spaces + "*")
  else: 
    inside_spaces = " " * (2*i-3)
    print(spaces + "*" + inside_spaces + "*")
#Bottom half
for i in range(row-1, 0, -1):
  spaces = " " * (row-i)
  if i==1:
    print(spaces + "*")
  else:
    inside_spaces = " " * (2*i-3)
    print(spaces+"*"+inside_spaces+"*")
```
