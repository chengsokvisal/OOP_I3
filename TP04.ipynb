- Create a class Book with title, author, and price attributes.
- Demonstrate encapsulation by protecting the price attribute.
- Use inheritance to create an EBook subclass.

```python
class book:
  def __init__(self, title, author, price):
    self.title = title
    self.author = author
    self.__price = price #Private attribute

  def get_price(self):
    return self.__price

  def set_price(self, price):
    if price>0:
      self.__price = price
    else:
      print("Price must be greater than 0")

  def display_info(self):
    print(f"Title: {self.title}")
    print(f"Author: {self.author}")
    print(f"Price: ${self.__price}")

#Using Inheritance
class EBook(book):
  def __init__(self, title, author, price, file_size):
    super().__init__(title, author, price)
    self.file_size = file_size

  def display_info(self):
    super().display_info()
    print(f"File size: {self.file_size} MB")    

book1 = book("Programming", "BOB", 25)
book1.display_info()

print()

# Create EBook object
ebook1 = EBook("Programming Advanced", "Jack", 15, 5)
ebook1.display_info()

print()

# Demonstrate Encapsulation
print("Book price:", book1.get_price())

book1.set_price(30)
print("Updated Price:", book1.get_price())

  
  
```
